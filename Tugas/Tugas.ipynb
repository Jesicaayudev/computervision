```python
import cv2
import numpy as np
import os

# ===== FUNCTION =====

def translasi(img, tx, ty):
    h, w = img.shape[:2]
    M = np.float32([[1, 0, tx], [0, 1, ty]])
    return cv2.warpAffine(img, M, (w, h))

def rotasi(img, sudut):
    h, w = img.shape[:2]
    center = (w // 2, h // 2)
    M = cv2.getRotationMatrix2D(center, sudut, 1)
    return cv2.warpAffine(img, M, (w, h))

def scaling(img, fx, fy):
    return cv2.resize(img, None, fx=fx, fy=fy)

def shearing(img):
    h, w = img.shape[:2]
    M = np.float32([[1, 0.5, 0],
                    [0, 1, 0]])
    return cv2.warpAffine(img, M, (w, h))

def refleksi(img):
    return cv2.flip(img, 1)

def affine(img):
    h, w = img.shape[:2]
    pts1 = np.float32([[50,50], [200,50], [50,200]])
    pts2 = np.float32([[10,100], [200,50], [100,250]])
    M = cv2.getAffineTransform(pts1, pts2)
    return cv2.warpAffine(img, M, (w, h))

def perspektif(img):
    h, w = img.shape[:2]
    pts1 = np.float32([[100,100], [400,100], [100,400], [400,400]])
    pts2 = np.float32([[50,150], [450,100], [150,450], [400,400]])
    M = cv2.getPerspectiveTransform(pts1, pts2)
    return cv2.warpPerspective(img, M, (w, h))


# ===== MAIN =====

input_folder = "/content/drive/MyDrive/Computer Vision/Tugas/dataset"
output_folder = "/content/drive/MyDrive/Computer Vision/Tugas/output"

# buat folder output kalau belum ada
os.makedirs(output_folder, exist_ok=True)

# filter hanya file gambar
files = [f for f in os.listdir(input_folder)
         if f.lower().endswith(('.jpg', '.jpeg', '.png'))]

print("Total file ditemukan:", len(files))

for file in files:
    path = os.path.join(input_folder, file)
    print("Proses:", file)

    img = cv2.imread(path)

    if img is None:
        print("Gagal load:", file)
        continue

    try:
        # grayscale
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

        # transformasi
        t = translasi(gray, 50, 30)
        r = rotasi(gray, 45)
        s = scaling(gray, 1.5, 1.5)
        sh = shearing(gray)
        rf = refleksi(gray)
        af = affine(gray)
        ps = perspektif(gray)

        # nama file aman
        nama = os.path.splitext(file)[0]

        # simpan
        cv2.imwrite(f"{output_folder}/{nama}_translasi.jpg", t)
        cv2.imwrite(f"{output_folder}/{nama}_rotasi.jpg", r)
        cv2.imwrite(f"{output_folder}/{nama}_scaling.jpg", s)
        cv2.imwrite(f"{output_folder}/{nama}_shearing.jpg", sh)
        cv2.imwrite(f"{output_folder}/{nama}_refleksi.jpg", rf)
        cv2.imwrite(f"{output_folder}/{nama}_affine.jpg", af)
        cv2.imwrite(f"{output_folder}/{nama}_perspektif.jpg", ps)

    except Exception as e:
        print("Error di file:", file, "|", str(e))

print("Selesai semua")
```
